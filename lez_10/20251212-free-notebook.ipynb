````python
import hashlib

class User:
    USER_ROLES = {"user", "administrator", "guest"}

    def __init__(self, username, role, password):
        if role not in self.USER_ROLES:
            raise ValueError(f"Invalid role: {role}. Must be one of {self.USER_ROLES}")
        if not isinstance(username, str):
            raise ValueError("Username must be a string")
        self._username = username
        self._role = role
        self._password_hash = self._hash(password)

    def _hash(self, plain: str) -> str:
        return hashlib.sha256(plain.encode()).hexdigest()
````

````python
class AdminUser(User):

    def __init__(self, username, password):
        super().__init__(username, 'administrator', password)
````

````python
user = AdminUser('ciccio', 'ciccio')
````

````python
class shape:
    def display(self):
        print("class shape")

class circle(shape):
    def display(self):
        print("class circle")

class square(shape):
    def display(self):
        print("class square")

class cylinder(square, circle):
    pass

obj = cylinder()
obj.display()
````

## The Diamond Problem

In multiple inheritance, the **diamond problem** occurs when a class inherits from two classes that share a common ancestor.

In our example:

```mermaid
graph TD
    shape --> circle
    shape --> square
    circle --> cylinder
    square --> cylinder
```

````python
cylinder.__mro__
````

````python
print("MRO:", [cls.__name__ for cls in cylinder.__mro__])
````

````python
class shape:
    def display(self):
        print("class shape")

class circle(shape):
    def display(self):
        print("class circle")
        super().display()

class square(shape):
    def display(self):
        print("class square")
        super().display()

class cylinder(circle, square):
    def display(self):
        print("class cylinder")
        super().display()

obj = cylinder()
obj.display()
````

````python
class NotAbstractDB:
    def connect(self):
        raise NotImplementedError("Subclasses should implement this")
        
    def disconnect(self):
        raise NotImplementedError("Subclasses should implement this")

class BadDB(NotAbstractDB):
    pass

db = BadDB() # Only fails when called — error is delayed
````

````python
db.connect() 
````

````python
from abc import ABC, abstractmethod

class AbstractDB(ABC):
    @abstractmethod
    def connect(self): ...
    
    @abstractmethod
    def disconnect(self): ...

class BadDB2(AbstractDB):
    pass

db = BadDB2()  # Raises TypeError immediately!
````
