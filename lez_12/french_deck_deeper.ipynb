```python
import collections

Card = collections.namedtuple('Card', 'rank suit') # namedtuples are class factory

# similar to
# class Card(object):
#     def __init__(self, rank, suit):
#         self.rank = rank
#         self.suit = suit
```

```python
SUITS = 'clubs diamonds spades hearts'.split()
RANKS = [str(n) for n in range(2, 11)] + list('JQKA')
CARDS = [Card(r, s) for s in SUITS for r in RANKS]
```

```python
class Deck:
    def __init__(self, cards=CARDS):
        self._cards = list(cards)
    # here we create a copy list this is to avoid sharing same reference
    # this is also safer with duck typing, I accept any iterable
    # there could be case in which you can't do this (i.e. copy of mln of elements)
    
    def __len__(self):
        return len(self._cards)
    
    def __getitem__(self, i):
        if isinstance(i, slice):
            return Deck(self._cards[i]) # this is a consistent choice, that is a new deck with sliced cards
        return self._cards[i]
    
    def __setitem__(self, i, card):
        self._cards[i] = card
    
    def __add__(self, other):
        return Deck(list(self) + list(other)) 
    # this work between a deck and another iterable 
    # that hopefully will be a list of Cards 
    # written like that it will not avoid or alert
    # "consenting adult"
    # convention -> arithmetic operator should build a new object not modify either of the arguments
    
    def deal(self, num_players):
        return [self[i::num_players] for i in range(num_players)]
    
```

```python
deck = Deck()
```

```python
top5 = deck[slice(0,5)] # same as deck[:5]
list(top5)
```
