# Polymorphism in Python
## What is Polymorphism?
Polymorphism in programming means the ability to present the same interface for differing underlying forms (data types or classes). The word comes from Greek poly (many) and morph (form) – literally "many forms"

In essence, a single function or method call can invoke different behaviors depending on the type of object it is operating on. This makes code more flexible and reusable, since you can write one piece of code to work with objects of different classes.

For example, a common real-world analogy is a remote control that can operate multiple devices (TV, AC, music system). You press the power button (one interface), and each device responds in its own way – the TV turns on, the AC starts cooling, the music system plays music. The interface is the same (power button), but the behavior differs based on the receiver (device). In programming, polymorphism allows the same function or method call to trigger different functionality depending on the object's type.


````python
class TV:
    def power(self):
        print("Turning on the TV")

class AirConditioner:
    def power(self):
        print("Starting the AC")

class MusicSystem:
    def power(self):
        print("Playing music")

# Common interface
def press_power(device):
    device.power()

# Each device behaves differently
press_power(TV())             # Output: Turning on the TV
press_power(AirConditioner()) # Output: Starting the AC
press_power(MusicSystem())    # Output: Playing music
````

## Static vs Dynamic Polymorphism (Compile-Time vs Run-Time)

Polymorphism can be categorized by when the method to execute is determined: at compile-time or at run-time. Static (compile-time) polymorphism means the decision of which method implementation to use is made during compilation. This typically includes method overloading or generic functions in statically-typed languages. It has no runtime overhead because the compiler chooses the correct method version before execution. Dynamic (run-time) polymorphism, on the other hand, defers the decision until the program is running. This is usually achieved via method overriding (sometimes called late binding or dynamic dispatch) where a call to a method like `obj.method()` is resolved to different implementations based on the actual type of obj at runtime. Dynamic polymorphism is more flexible (it even enables Python’s duck typing) but can be slightly slower due to the runtime method look-up.

Python is a dynamically-typed language, so it leans **heavily on runtime polymorphism**. Unlike C++ or Java, Python doesn’t perform method binding at compile time – there is in fact no true compile-time method overloading in Python. Every method call is resolved at run-time, which gives Python its flexibility. (Note: Built-in operators like + are still polymorphic, but the language itself doesn’t allow user-defined overloaded function signatures at compile time)

````python
# Simulated overloading via default arguments
def greet(name=None):
    if name:
        print(f"Hello, {name}!")
    else:
        print("Hello!")

greet()         # Output: Hello!
greet("Alice")  # Output: Hello, Alice!
````

In statically-typed languages you would define:
```java
void greet() { ... }
void greet(String name) { ... }
```

````python
class Animal:
    def speak(self):
        print("Some sound")

class Dog(Animal):
    def speak(self):
        print("Woof!")

class Cat(Animal):
    def speak(self):
        print("Meow!")

def make_it_speak(animal: Animal):
    animal.speak()

make_it_speak(Dog())  # Output: Woof!
make_it_speak(Cat())  # Output: Meow!
````

## Ad Hoc vs Parametric Polymorphism (and Subtype Polymorphism)

Another theoretical way to classify polymorphism is by the form of implementation:

- **Ad hoc polymorphism**: This refers to defining multiple implementations of the same function name tailored for different types. Examples include function overloading (having several functions with the same name but different parameter types) and operator overloading (the `+` operator adding numbers or concatenating strings). The appropriate implementation is chosen per type, often at compile time in statically-typed languages. (In Python, ad hoc polymorphism appears in the form of dynamic dispatch and special methods for operators, since Python doesn’t do compile-time overload resolution)

- **Parametric polymorphism**: This is writing code in a generic manner such that it can operate on any type (often using type parameters). For instance, a single sort function that can sort a list of `T` for any type `T` is parametric. In object-oriented terms, this is achieved via generics (like Java generics or C++ templates) – one function/class works for all types without modification. The implementation is agnostic to type and typically determined at compile time by the compiler generating a version for each needed type (in static languages). Python doesn’t need explicit generics for this – its dynamic typing means many functions are automatically generic (e.g. the built-in `len()` works on multiple container types).

- **Subtype polymorphism**: Sometimes just called inclusion polymorphism, this is the classic OOP polymorphism via inheritance – a base class reference can refer to an object of a subclass, and methods overridden by the subclass will be called. This is always resolved at runtime (dynamic dispatch) because the exact subtype isn’t known until then. For example, a base class Animal might have a method `speak()`, and subclasses Dog and Cat override it; if you write code against an Animal interface, at runtime a Dog or Cat object will produce different results for `speak()`.

These categories often align with the static/dynamic distinction: Static polymorphism typically occurs with ad hoc and parametric polymorphism (decided by the compiler), while subtype polymorphism is achieved via dynamic (runtime) polymorphism. Python primarily uses subtype polymorphism (via method overriding and duck typing) and ad-hoc polymorphism at runtime, given its dynamic nature.

### Ad Hoc Polymorphism in Python

Python achieves ad hoc polymorphism through dynamic dispatch and operator overloading (via special methods):

Operator Overloading Example:
```python
print(3 + 5)          # int + int → 8
print("a" + "b")      # str + str → "ab"
print([1, 2] + [3])   # list + list → [1, 2, 3]

```
Here, the same + operator behaves differently depending on the operand types. Behind the scenes, Python dispatches to `__add__()` of each type.

Dynamic Dispatch Example:
```python
def add(x, y):
    return x + y

add(1, 2)           # 3
add("hello", "hi")  # "hellohi"
add([1], [2, 3])    # [1, 2, 3]
```

This is ad hoc polymorphism resolved at runtime, because Python determines which `+` implementation to invoke based on operand types.

### Parametric Polymorphism in Python

In statically typed languages, parametric polymorphism looks like this:
```java
public class Box<T> {
    T value;
    public Box(T value) { this.value = value; }
    public T get() { return value; }
}
```

In Python, dynamic typing provides implicit parametric polymorphism:
```python
def identity(x):
    return x

print(identity(42))         # 42
print(identity("string"))   # "string"
print(identity([1, 2, 3]))  # [1, 2, 3]
```

The same function works with any type — no type declaration needed. Python's identity here is generic.

Another example:
```python
def double_each(items):
    return [item * 2 for item in items]

double_each([1, 2])       # [2, 4]
double_each(["a", "b"])   # ["aa", "bb"]
```

double_each works for any list of elements that support `* 2`, without being rewritten for each type.

### Subtype Polymorphism (Inheritance and Overriding)

This is the classical OOP polymorphism:
```python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

def let_it_speak(animal: Animal):
    print(animal.speak())

let_it_speak(Dog())   # Woof
let_it_speak(Cat())   # Meow
```

Here, let_it_speak accepts any Animal, but calls the `speak()` method that is overridden in each subclass. This is resolved at runtime.

## Polymorphism in Python: Key Mechanisms

Python’s dynamic nature means polymorphism is deeply integrated into how you write Python code. Here are the key polymorphic mechanisms in Python:

Duck Typing: “If it walks like a duck and quacks like a duck, it must be a duck.” Duck typing means the type of an object is less important than the methods or attributes it defines. In Python, you don’t need an explicit interface or base class for polymorphism – you just call methods assuming the object supports them. For example, if you have a function that calls `obj.fly()`, it will work with any object that implements a `.fly()` method (whether it’s a Duck class or an Airplane class). This allows writing very flexible code: as long as an object “quacks” (supports the expected method), it can be used in place of any other with the same behavior. Duck typing is a form of polymorphism specific to dynamically-typed languages like Python, Ruby, etc., and it relies on runtime method resolution.

**Method Overriding (Subtype Polymorphism)**: Python fully supports inheritance and overriding methods in subclasses. A subclass can provide a specific implementation of a method that is already defined in its parent class. When you call that method on an instance of the subclass, Python dynamically dispatches to the subclass’s version. For example, you might have:
```python
class Animal:
    def speak(self):
        return "...?..."
class Cat(Animal):
    def speak(self):
        return "Meow"
class Dog(Animal):
    def speak(self):
        return "Bark"
```

Here, Cat and Dog override the speak() method of Animal. If you iterate a list of Animal objects that actually contains a Cat and a Dog, each will respond to speak() with its own sound. This is runtime polymorphism – the same method call `animal.speak()` exhibits different behavior depending on the actual subclass of animal.

**Operator Overloading**: Many Python operators are polymorphic by design. The + operator is a classic example: `5 + 10` invokes integer addition, whereas "Hello " + "World!" invokes string concatenation, and `[1,2] + [3,4]` merges lists. When you define your own classes, you can overload operators by defining special methods like `__add__`, `__sub__`, `__str__`, etc. For instance, you could define a Point class and implement `__add__` so that `point1 + point2` yields a new Point with summed coordinates. This ability to use the same operator symbol for different types is a form of ad hoc polymorphism. Under the hood, Python decides which method to call based on the types of the operands at runtime (there’s a different implementation of `__add__` for numbers, strings, lists, and so on).

Function Overloading via Default Arguments and Single Dispatch: Python does not support defining multiple functions with the same name but different signatures (as languages like C++ or Java do). However, you can achieve similar flexibility in two ways:

- **Default/Variable Arguments**: You can define a function with optional parameters or `*args`/`**kwargs` to handle various cases. For example, you might write a single function `def area(length, width=None)` that computes area differently depending on whether width is provided (treating it as either a square or rectangle). This way, `area(5)` and `area(5, 3)` both work, acting like two overloads of an area function. In object methods, you can similarly use default parameters to allow flexible calls. While not true compile-time overloading, it achieves a polymorphic behavior (the function adapts to different argument patterns) in a simple manner.

- **Single Dispatch Generic Functions**: The `functools.singledispatch` decorator in Python’s standard library allows you to register multiple versions of a function for different argument types. This is a form of dynamic function overloading. For example, you could define a generic function `@singledispatch def render(obj)` and then provide specialized versions for, say, `@render.register(Point)` or `@render.register(Circle)` to handle different types. At runtime, the correct version is chosen based on the type of the first argument. This provides a clean way to implement ad hoc polymorphism (functions that behave differently on different argument types) without littering the code with type checks or if/elif chains.


### Duck Typing
```python
class Duck:
    def fly(self):
        print("Flap flap!")

class Airplane:
    def fly(self):
        print("Jet engines roaring!")

def launch_into_air(flyer):
    flyer.fly()  # Works as long as object has a .fly() method

launch_into_air(Duck())      # Flap flap!
launch_into_air(Airplane())  # Jet engines roaring!
```

The function doesn’t care what the type is — only that .fly() exists. This is duck typing.

### Method Overriding (Subtype Polymorphism)
```python
class Animal:
    def speak(self):
        return "...?..."

class Cat(Animal):
    def speak(self):
        return "Meow"

class Dog(Animal):
    def speak(self):
        return "Bark"

animals = [Cat(), Dog()]

for a in animals:
    print(a.speak())  # Output: Meow, then Bark
```

Here, the same speak() method behaves differently based on the subclass at runtime.

### Operator Overloading
```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1 + p2)  # (4, 6)
```

This is polymorphic use of +, which invokes __add__ depending on the type.

### Function Overloading via Default Arguments
```python
def area(length, width=None):
    if width is None:
        return length * length  # square
    return length * width       # rectangle

print(area(5))      # 25
print(area(5, 3))   # 15
```

A single function adapts to different signatures — a form of ad hoc polymorphism.

### Function Overloading via Single Dispatch
```python
from functools import singledispatch

@singledispatch
def render(obj):
    print("Default rendering")

@render.register
def _(obj: int):
    print(f"Rendering integer: {obj}")

@render.register
def _(obj: str):
    print(f"Rendering string: '{obj}'")

render(10)       # Rendering integer: 10
render("hello")  # Rendering string: 'hello'
render(3.14)     # Default rendering
```

Each version of render() is dispatched at runtime based on the argument type.

## Polymorphism in Python vs. Java/C++

To put Python’s approach in context, here’s a brief comparison with static languages like Java or C++:

**Method Overloading**: Languages like Java and C++ support compile-time polymorphism via method overloading (multiple methods same name, different parameter types). Python lacks this – you cannot define two methods with the same name in one scope. Attempts to do so will simply rebind the name to the last definition. Instead, Python encourages using default parameters or single-dispatch as described above.

**Parametric Polymorphism (Generics)**: Java and C++ use generics/templates to achieve type-agnostic code (parametric polymorphism) at compile time. Python’s equivalent is simply to write functions and classes that assume operands support the needed operations (duck typing). For example, a Java method might be generic List<T> sort(List<T> list), whereas in Python you’d write a sort(list) function that works for any list of objects that can be ordered, relying on those objects to implement the comparison methods. Python’s upcoming type hints for generics are only for static type checking, not for altering runtime behavior. In practice, duck typing gives Python code a lot of the flexibility that generics give to Java, but without explicit syntax.

**Inheritance and Interfaces**: Java requires classes to declare implemented interfaces or extend base classes to use them polymorphically, and C++ requires virtual functions for runtime polymorphism. Python has no formal interface declarations – any object that has the needed method can be used. You can still use abstract base classes (abc.ABC) in Python to define a formal interface, but it's not required. This makes Python polymorphism more implicit – it’s based on object capabilities rather than object types. In Java, if you have a function that takes an interface Drawable, only objects of classes implementing Drawable can be passed. In Python, if you have a function that calls obj.draw(), any object with a draw method will work, regardless of its class.

**Runtime Dispatch Performance**: In C++, if you use templates (parametric polymorphism) or function overloading, the decisions are made at compile time (zero runtime cost for dispatch). If you use virtual functions (subtype polymorphism), there’s a v-table lookup at runtime (small cost). Python, being dynamically typed, always uses runtime dispatch for method calls (essentially similar to a virtual function lookup, or a dictionary lookup for the attribute). This is inherently a bit slower than a direct function call in C++, but it’s the price of flexibility. As noted in theory, “dynamic polymorphism is more stable, but slower” – in Python this trade-off is generally acceptable given modern computing power and the benefits of write-time flexibility.
