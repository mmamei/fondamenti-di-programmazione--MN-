## Duck Typing at his best!

```python
class AddressList:
    def __init__(self, entries=None):
        self._entries = []
        if entries:
            for entry in entries:
                self._validate_entry(entry)
                self._entries.append(entry)

    def _validate_entry(self, value):
        if not isinstance(value, dict):
            raise TypeError("Entry must be a dictionary.")
        if not {"name", "number"}.issubset(value.keys()):
            raise ValueError("Entry must have both 'name' and 'number' keys.")

    def __getitem__(self, index):
        return self._entries[index]

    def __setitem__(self, index, value):
        self._validate_entry(value)
        self._entries[index] = value

    def __len__(self):
        return len(self._entries)

    def __add__(self, other):
        if not isinstance(other, AddressList):
            raise TypeError("Can only add another AddressList")
        return AddressList(self._entries + other._entries)

    def append(self, value):
        self._validate_entry(value)
        self._entries.append(value)

    def __repr__(self):
        return f"AddressList({self._entries})"
```

```python
# Create two address lists
a1 = AddressList([
    {"name": "Alice", "number": "1234"},
    {"name": "Bob", "number": "5678"},
])

a2 = AddressList([
    {"name": "Charlie", "number": "9999"}
])
```

```python
print(a1)
```

```python

# Access like a list
print(a1[0])  # {'name': 'Alice', 'number': '1234'}


```

```python
# Set like a list
a1[1] = {"name": "Bobby", "number": "7777"}
```

```python

# Append new entries
a1.append({"name": "Daisy", "number": "3333"})

```

```python
# Add two address lists
combined = a1 + a2
```

```python
print(combined)
print(len(combined)) 
```
