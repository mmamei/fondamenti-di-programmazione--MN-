```python
def leggi(nome_file):
    '''
    devi leggere il file a/esame1.txt
    il file contiene informazioni su gruppi di liste di numeri
    Ad esempio:

    Nome: A

    Valori: 1,2,3,4,5

    Nome: B
    Valori: 3,4,5

    Un dato gruppo può comparire più volte se quel gruppo contiene più liste.


    Leggi il file e organizza le informazini in un dizionario che associa i nomi dei gruppi
    a delle liste di liste di valori

    Crea la funzione e restituisci il dizionario

    >>> leggi('a/esame1.txt')
    {'A': [[1, 2, 3, 4, 5], [4, 2]], 'B': [[3, 4, 5]], 'C': [[1, 2, 3, 6]]}

    >>> leggi('a/esame2.txt')
    {'C': [[1, 2, 3, 6]]}

    '''
    gruppi = {}
    nome_gruppo = None

    with open(nome_file, "r", encoding="utf-8") as file:
        for riga in file:
            riga = riga.strip()

            if riga.startswith("Nome:"):
                nome_gruppo = riga.split(":")[1].strip()
                if nome_gruppo not in gruppi:
                    gruppi[nome_gruppo] = []
            elif riga.startswith("Valori:") and nome_gruppo is not None:
                testo_valori = riga.split(":", 1)[1]
                valori = [int(valore.strip()) for valore in testo_valori.split(",")]
                gruppi[nome_gruppo].append(valori)

    return gruppi


def gurppi_con_almeno_un_vettore_pari(gruppi):
    '''
    Scrivi una funzione che prende in input il dizionario dei gruppi e restituisce una lista
    con i nomi dei gruppi che contengono almeno un vettore con tutti valori pari
    non usare comandi come all, any, map, filter, reduce, ecc. ma solo cicli for e if

    >>> diz = {'A': [[1, 2, 3, 4, 5], [2, 4, 6]], 'B': [[3, 4, 5]], 'C': [[2, 4], [1, 3]]}
    >>> gurppi_con_almeno_un_vettore_pari(diz)
    ['A', 'C']

    >>> diz = {'A': [[1, 2, 3, 4, 5]], 'B': [[3, 4, 5]], 'C': [[1, 3]]}
    >>> gurppi_con_almeno_un_vettore_pari(diz)
    []

    '''
    gruppi_pari = []

    for nome_gruppo, liste_valori in gruppi.items():
        for valori in liste_valori:
            is_pari = True
            for valore in valori:
                if valore % 2 != 0:
                    is_pari = False
                    break
            if is_pari:
                gruppi_pari.append(nome_gruppo)
                break
    return gruppi_pari


import doctest
doctest.testmod(verbose=False)


diz = leggi("a/esame1.txt")
print(diz)
print(gurppi_con_almeno_un_vettore_pari(diz))
```
