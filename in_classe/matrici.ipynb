```python
x = [
    [
        [1,2,3],
        [4,5,6],
        [7,8,9]
    ],
    [[1,2],[0,4]],
    [[7,7],[7,7]],
    [[300]]
]

def sum_m(m):
    #s = 0
    #for r in m:
    #    s += sum(r)
    #return s
    return sum([sum(r) for r in m])

somme_x = [sum_m(m) for m in x]
i_max = somme_x.index(max(somme_x))
print(x[i_max])

```

```python
def sum_riga(m,i):
    return sum(m[i])

def sum_colonna(m,j):
    s = 0
    for r in m:
        s += r[j]
    return s

def trova_riga_massima(m):
    somme_r = [sum_riga(m,i) for i in range(len(m))]
    i_max = somme_r.index(max(somme_r))
    return i_max

def trova_colonna_massima(m):
    somme_c = [sum_colonna(m,j) for j in range(len(m[0]))]
    j_max = somme_c.index(max(somme_c))
    return j_max

def print_colonna(m,j):
    for r in m:
        print(r[j])


for m in x:
    i_max = trova_riga_massima(m)
    j_max = trova_colonna_massima(m)
    print(m[i_max])
    print_colonna(m,j_max)
    print('---------------------')
```

```python
def trova_dimensione(m):
    return str(len(m))+'X'+str(len(m[0]))

def crea_dizionario(m):
    diz = {}
    for m in x:
        dim = trova_dimensione(m)
        if dim not in diz:
            diz[dim] = [m]
        else:
            diz[dim].append(m)
    return diz


def somma_matriciale(x):
    ris = []
    for r in x[0]:
        ris.append([0 for _ in r])

    for m in x:
        for i in range(len(m)):
            for j in range(len(m[0])):
                ris[i][j] += m[i][j]
    return ris


diz = crea_dizionario(x)
for dim, matrici in diz.items():
    ris = somma_matriciale(matrici)
    print(dim,'-->',ris)


```
