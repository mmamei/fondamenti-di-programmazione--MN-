```python
# leggi un numero n
# leggi n numeri
# calcola la somma degli n numeri letti

n = int(input("Inserisci un numero: "))
somma = 0
for i in range(n):
    x = int(input("Inserisci un numero: "))
    somma += x
print("La somma dei numeri inseriti è:", somma)
```
