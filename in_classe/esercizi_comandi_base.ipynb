```python
# leggi un numero n
# leggi n numeri
# calcola la somma degli n numeri letti

n = int(input("Inserisci un numero: "))
somma = 0
for i in range(n):
    x = int(input("Inserisci un numero: "))
    somma += x
print("La somma dei numeri inseriti è:", somma)
```

```python
x = 5
a = []
for i in range(x):
    a.append(int(input('inserisci numero ')))

print(a)

media = 0 
for x in a:
    media += x
media /= len(a)

print('media',media)

sd = 0
for x in a:
    sd += (x - media)**2
sd /= len(a)
sd = sd ** 0.5

print('deviazione standard', sd)
```

```python


# cerca un elemento x dentro a una lista b
# se lo trova ritorna True altrimenti ritorna False

def trova(x,b):
    trovato = False
    for z in b:
        if x == z:
            trovato = True
            break
    return trovato


# elimina i duplicati da una lista a 
# ritorna una lista senza duplicati

def elimina_duplicati(a):
    b = []
    for x in a:
        if not trova(x,b): 
            b.append(x)
    return b

# dato un elemento y e una lista a
# conta quante volte y compare nella lista a

def calcola_frequenza(y, a):
    sum = 0
    for x in a:
        if x == y:
            sum += 1
    return sum


a = [1,2,1,3,2,1,4,8]
b = elimina_duplicati(a)
print(b)
for y in b:
    f = calcola_frequenza(y,a)
    print(y,'----->',f)
```
