```python
N = 5

for i in range(N):
    for j in range(N):
        print("*", end="  ")
    print("")
```

```python
N = 9

for i in range(N):
    for j in range(N):
        if i == j or j == N - 1 - i:
            print("*", end="  ")
        else:
            print(".", end="  ")
    print("")
```
